```python
import csv
def load_dataset(filename):
    rows = []
    with open(filename, mode='r', encoding = "utf-8", newline= "") as file:
        reader = csv.DictReader(file)
        for row in reader:
            rows.append(row)

    return rows
```

```python
transactions = load_dataset("./data/data_ecommerce_customer_churn.csv")

print("Number of rows:", len(transactions))
print("First row:")
print(transactions[0])

print("\nColumns:")
print(transactions[0].keys())
```

```python
numerical_columns = [
    "Tenure",
    "WarehouseToHome",
    "NumberOfDeviceRegistered",
    "SatisfactionScore",
    "NumberOfAddress",
    "Complain",
    "DaySinceLastOrder",
    "CashbackAmount"
]
target_column = "Churn"
```

```python
import numpy as np
```

```python
print("Columns:")
for column in transactions[0].keys():
    print(repr(column))
```

```python
def transaction_clean(transactions):
    clean_transactions = []
    for row in transactions:
        valid = True
        for column in numerical_columns:
            if row[column].strip() == "":
                valid = False
                break
        if valid:
            clean_transactions.append(row)

    print("Original rows:", len(transactions))
    print("Clean rows:", len(clean_transactions))
    print("Removed rows:", len(transactions) - len(clean_transactions))
    return clean_transactions
```

```python
X = np.array([
    [
        float(row[column])
        for column in numerical_columns
    ]
    for row in transaction_clean(transactions)
])

print("X shape:", X.shape)
print("X ndim:", X.ndim)
print("X size:", X.size)
print("X dtype:", X.dtype)
```
