```python
from statistics import median


def load_dataset(filename):
    import csv
    rows = []
    with open(filename, 'r' , encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            rows.append(row)

    return rows
```

```python
numerical_columns = [
    'Tenure',
    'WarehouseToHome',
    'NumberOfDeviceRegistered',
    'SatisfactionScore',
    'NumberOfAddress',
    'Complain',
    'DaySinceLastOrder',
    'CashbackAmount'
]
```

```python
import numpy as np
```

```python
def transaction_clean(transactions):
    clean_transactions = []
    for row in transactions:
        valid = True
        for column in row:
            if row[column].strip() == "":
                valid = False
                break
        if valid:
            clean_transactions.append(row)
    return clean_transactions
```

```python
transactions = load_dataset('data/data_ecommerce_customer_churn.csv')
clean_transactions = transaction_clean(transactions)
```

```python
X = np.array([
    [float(row[column]) for column in numerical_columns]
    for row in clean_transactions
])
print("X shape:", X.shape)
print("X ndim:", X.ndim)
print("X size:", X.size)
print("X dtype:", X.dtype)
```

```python
low_satisfaction = X[:, 3] <= 2
has_complaint = X[:, 5] == 1
inactive = X[:, 6] > 10
low_tenure = X[:, 0] < 5
few_addresses = X[:, 4] <= 1

risk_score = (
    np.where(low_satisfaction,2,0)+
    np.where(has_complaint,2,0)+
    np.where(inactive,2,0)+
    np.where(low_tenure,1,0)+
    np.where(few_addresses,1,0)
)

print("Minimum risk:", np.min(risk_score))
print("Maximum risk:", np.max(risk_score))

print("\nFirst 20 risk scores:")
print(risk_score[:20])
```

```python
risk_group = np.where(
    risk_score <=2 , "Low",
    np.where(risk_score <= 4 , "Medium",
             "High"
             )
)
print(risk_group[:20])

print()
print(np.unique(risk_group,return_counts=True))
```

```python
low_risk = risk_group == "Low"
low_risk_total  = np.sum(low_risk)
low_risk_churned = np.sum(y[low_risk] == 1)

low_churn_rate = low_risk_churned / low_risk_total
print("Low Risk")
print("Customers:", low_risk_total)
print("Churned:", low_risk_churned)
print("Churn rate:", low_churn_rate * 100, "%")
```

```python
medium_risk = risk_group == "Medium"

medium_risk_total = np.sum(medium_risk)
medium_risk_churn = np.sum(y[medium_risk] == 1)
medium_churn_rate = medium_risk_churn/medium_risk_total

print("Medium Risk")
print("Customers:", medium_risk_total)
print("Churned:", medium_risk_churn)
print("Churn rate:", medium_churn_rate * 100, "%")
```

```python
high_risk = risk_group == "High"
high_risk_total = np.sum(high_risk)
high_risk_churn = np.sum(y[high_risk] == 1)

high_churn_rate = high_risk_churn/high_risk_total

print("High Risk")
print("Customers:", high_risk_total)
print("Churned:", high_risk_churn)
print("Churn rate:", high_churn_rate * 100, "%")
```

```python
import numpy as np
import time

values = np.random.random(1_000_000)

start = time.time()

result_loop = []

for x in values:
    result_loop.append(x**2 + 2*x + 5)

loop_time = time.time() - start

print("Python loop time:", loop_time)

start = time.time()

result_numpy = values**2 + 2*values + 5

numpy_time = time.time() - start

print("NumPy time:", numpy_time)

speedup = loop_time / numpy_time

print("Python loop:", loop_time)
print("NumPy:", numpy_time)
print("Speedup:", speedup, "x")
```
