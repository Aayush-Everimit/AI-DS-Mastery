```python
import csv


def load_dataset(filename):
    rows = []
    with open(filename, "r" , encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            rows.append(row)

    return rows
```

```python
data = dict()
data = load_dataset('data/train.csv')
print("FIRST ROW \n",data[:1])
```

```python
Identifiers = [
    "Row ID",
    "Order ID",
    "Customer ID",
    "Product ID"
]

Dates = [
    "Order Date",
    "Ship Date"
]

Categorical = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Postal Code",
    "Category",
    "Sub-Category",
    "Product Name"
]

Numerical = [
    "Sales"
]
```

```python
def transaction_clean(transactions):
    clean_transactions = []

    for row in transactions:
        valid = True

        for column in row:
            if str(row[column]).strip() == "":
                valid = False
                break

        if not valid:
            continue
        for col in Numerical:
            if col in row:
                try:
                    value = float(row[col])
                    if value < 0:
                        valid = False
                        break
                except (ValueError, TypeError):
                    valid = False
                    break
            else:
                valid = False
                break

        if valid:
            clean_transactions.append(row)

    return clean_transactions
```

```python
working_data = transaction_clean(data)
```
