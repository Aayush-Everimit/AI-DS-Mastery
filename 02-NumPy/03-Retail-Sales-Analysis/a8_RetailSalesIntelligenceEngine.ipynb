```python
import csv


def load_dataset(filename):
    rows = []
    with open(filename, "r" , encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            rows.append(row)

    return rows
```

```python
data = dict()
data = load_dataset('data/train.csv')
print("FIRST ROW \n",data[:1])
```

```python
Identifiers = [
    "Row ID",
    "Order ID",
    "Customer ID",
    "Product ID"
]

Dates = [
    "Order Date",
    "Ship Date"
]

Categorical = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Postal Code",
    "Category",
    "Sub-Category",
    "Product Name"
]

Numerical = [
    "Sales"
]
```

```python
def transaction_clean(transactions):
    clean_transactions = []

    for row in transactions:
        valid = True

        for column in row:
            if str(row[column]).strip() == "":
                valid = False
                break

        if not valid:
            continue
        for col in Numerical:
            if col in row:
                try:
                    value = float(row[col])
                    if value < 0:
                        valid = False
                        break
                except (ValueError, TypeError):
                    valid = False
                    break
            else:
                valid = False
                break

        if valid:
            clean_transactions.append(row)

    return clean_transactions
```

```python
import datetime
import numpy as np

date_format = "%d/%m/%Y"

shipping_days = list()
for transaction in working_data:
    order_date = datetime.datetime.strptime(transaction["Order Date"], date_format)
    ship_date = datetime.datetime.strptime(transaction["Ship Date"], date_format)
    duration = ship_date - order_date
    shipping_days.append(duration.days)


X = np.array(shipping_days)
print("Average shipping days : ", np.mean(X))
print("Median shipping days : ", np.median(X))
print("Minimum shipping days : " , np.min(X))
print("Maximum shipping days : " , np.max(X))

print()
print("Shipping Duration Distribution:")
same_day = np.sum(X == 0)
one_two_days = np.sum((X >= 1) & (X <= 2))
three_four_days = np.sum((X >= 3) & (X <= 4))
five_plus_days = np.sum(X >= 5)

print("0 days:", same_day)
print("1-2 days:", one_two_days)
print("3-4 days:", three_four_days)
print("5+ days:", five_plus_days)
```

```python
ship_mode_shipping_days = {}
for transaction , days in zip(working_data, shipping_days):
    ship_mode = transaction["Ship Mode"]
    if ship_mode not in ship_mode_shipping_days :
        ship_mode_shipping_days[ship_mode] = []

    ship_mode_shipping_days[ship_mode].append(days)



for ship_mode , days in ship_mode_shipping_days.items():
    mode_days = np.array(days)

    print("Ship Mode:", ship_mode)
    print("Transaction Count =", len(mode_days))
    print("Average Shipping Days =", np.mean(mode_days))
    print("Median Shipping Days =", np.median(mode_days))
    print("Maximum Shipping Days =", np.max(mode_days))
    print()
```
