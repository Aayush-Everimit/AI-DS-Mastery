```python
def get_unique_transaction_types(transactions):
    unique_transaction_types=set()
    for transaction in transactions:
        unique_transaction_types.add(transaction.get("type"))

    return unique_transaction_types
```

```python
def group_transactions_by_origin_account(transactions):
    transactions_by_origin_account = {}

    for transaction in transactions:
        origin_account = transaction.get("nameOrig")

        if origin_account not in transactions_by_origin_account:
            transactions_by_origin_account[origin_account] = []

        transactions_by_origin_account[origin_account].append(transaction)

    return transactions_by_origin_account
```

```python
def find_accounts_with_multiple_transactions(account_transactions):
    accounts_with_multiple_transactions = set()

    for account, count in account_transactions.items():
        if len(count) > 1:
            accounts_with_multiple_transactions.add(account)

    return accounts_with_multiple_transactions
```

```python
def count_transactions_by_type(transactions):
    transactions_freq_by_type = {}
    for transaction in transactions :
        transaction_type = transaction.get("type")
        if transaction_type not in transactions_freq_by_type:
            transactions_freq_by_type[transaction_type] = 1
        else:
            transactions_freq_by_type[transaction_type] += 1

    return transactions_freq_by_type
```

```python
def get_most_common_transaction_type(transactions):
    transactions_frequency = count_transactions_by_type(transactions)
    sorted_transactions_frequency = sorted(transactions_frequency.items(), key=lambda item: item[1], reverse=True)
    return sorted_transactions_frequency[0][0]
```

```python
def find_duplicate_transactions(transactions):
    unique_transactions = set()
    duplicate_transaction = set()

    for transaction in transactions:
        transaction_id = (
            transaction.get("step"),
            transaction.get("type"),
            transaction.get("amount"),
            transaction.get("nameOrig"),
            transaction.get("nameDest")
        )
        if transaction_id not in unique_transactions:
            unique_transactions.add(transaction_id)

        else:
            duplicate_transaction.add(transaction_id)


    return duplicate_transaction
```

```python
def calculate_origin_account_spending(transactions):
    origin_account_spending = {}
    for transaction in transactions:
        origin_account = transaction.get('nameOrig')
        if origin_account not in origin_account_spending:
            origin_account_spending[origin_account] = float(0)
        amount = float(transaction.get('amount'))
        origin_account_spending[origin_account] += amount

    return origin_account_spending
```

```python
def get_top_spending_accounts(transactions, k):
    top_spending_accounts = sorted(calculate_origin_account_spending(transactions).items() , key=lambda item: item[1] , reverse = True)[:k]
    return top_spending_accounts
```

```python
def count_fraud_by_type(transactions):
    transactions_freq_type = {}
    for transaction in transactions:
        transaction_type = transaction.get("type")
        if transaction_type not in transactions_freq_type:
            if transaction['isFraud'] == 1:
                transactions_freq_type[transaction_type] = 1
            else:
                transactions_freq_type[transaction_type] = 0
        else:
            if transaction['isFraud'] == 1:
                transactions_freq_type[transaction_type] += 1
            else:
                transactions_freq_type[transaction_type] += 0

    return transactions_freq_type
```

```python
def get_fraudulent_accounts(transactions):
    fraudulent_accounts = set()
    for transaction in transactions:
        origin_account = transaction.get('nameOrig')
        if transaction['isFraud'] == 1:
            fraudulent_accounts.add(origin_account)

    return fraudulent_accounts
```

```python
# transaction_id = f"{transaction.get('step')}_{transaction.get('type')}_{transaction.get('amount')}_{transaction.get('nameOrig')}_{transaction.get('nameDest')}"

def build_account_lookup(transactions):
    account_lookup = {}

    for transaction in transactions:
        account_id = transaction.get("nameOrig")

        if account_id not in account_lookup:
            account_lookup[account_id] = []

        account_lookup[account_id].append(transaction)

    return account_lookup


def get_account_transactions(account_lookup, account_id):
    return account_lookup.get(account_id, [])
```
