```python
def validate_name(name):
    if type(name) == str :
        return True
    else:
        return False
```

```python
def validate_age(age):
    if type(age) == int and 120 > age >= 0:
        return True
    else:
        return False
```

```python
def validate_marks(marks):
    return isinstance(marks, (int, float)) and 0 <= marks <= 100
```

```python
def clean_student(student):
    if student is None :
        return False,"invalid student"
    if student["name"] == "" :
        return False,"invalid name"
    if student["age"] == "":
        return False,"invalid age"
    if student["marks"] == "":
        return False,"invalid marks"



    if not validate_name(student["name"]) :
        student["name"] = str(student["name"]).strip().title()

    if not validate_age(student["age"]):
        student["age"] = int(student["age"])
        if student["age"] < 0 : return False,"invalid age, less than 0"

    if not validate_marks(student["marks"]):
        student["marks"] = float(student["marks"])
        if 0 <= student["marks"] <= 100 : return False,"invalid marks, not in range!"


    return student
```

```python
def calculate_average(students):
    if not students:
        return None
    count = 0
    marks = 0
    for student in students :
        marks += student["marks"]
        count += 1

    average = marks/count

    return average
```

```python
def find_highest_marks(students):
    if not students:
        return None
    highest = 0.00
    for student in students :
        if student["marks"] > highest : highest = student["marks"]
    return highest
```

```python
def find_lowest_marks(students):
    if not students:
        return None
    lowest = 100.00
    for student in students :
        if student["marks"] < lowest : lowest = student["marks"]

    return lowest
```

```python
def calculate_median(students):
    if not students:
        return None
    marks = []
    for student in students:
        marks.append(student["marks"])
    marks.sort()


    n = len(marks)
    if n % 2 == 1:
        median = marks[n // 2]
    else:
        middle1 = marks[n // 2 - 1]
        middle2 = marks[n // 2]
        median = (middle1 + middle2) / 2

    return median
```

```python
def filter_students(students, min_marks):
    if not students:
        return None
    filtered = []
    for student in students :
        if student["marks"] >= min_marks :
            filtered.append(student)

    return filtered
```

```python
def assign_grade(marks):
    if 90 <= marks <= 100:
        grade = "A"
    elif 80 <= marks < 90:
        grade = "B"
    elif 70 <= marks < 80:
        grade = "C"
    elif 60 <= marks < 70:
        grade = "D"
    else:
        grade = "E"

    return marks, grade
```

```python
def summary(students, passing_marks = 40):
    cleaned_students = []
    for student in students:
        cleaned = clean_student(student)
        if isinstance(cleaned, tuple) and not cleaned[0]:
            continue
        cleaned_students.append(cleaned)
    if not cleaned_students:
        return {"error": "No valid student data found"}

    passed_count = 0
    failed_count = 0
    grade_distribution = {"A" : 0 , "B" : 0 , "C" : 0 , "D" : 0 , "E" : 0 }
    for student in cleaned_students :
        if student["marks"] >= passing_marks :
            passed_count += 1
        else :
            failed_count += 1

        _,grade = assign_grade(student["marks"])
        if grade in grade_distribution:
            grade_distribution[grade] += 1
    return {
        "Number of students": len(cleaned_students),
        "Average marks": calculate_average(cleaned_students),
        "Highest marks": find_highest_marks(cleaned_students),
        "Lowest marks": find_lowest_marks(cleaned_students),
        "Median marks": calculate_median(cleaned_students),
        "Number passed": passed_count,
        "Number failed": failed_count,
        "Grade distribution": grade_distribution
    }
```
