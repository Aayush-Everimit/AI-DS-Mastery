```python
import csv
```

```python
def load_dataset(file_name):
    rows = []

    with open(file_name, "r", encoding="latin-1") as file:
        csv_reader = csv.DictReader(file)

        for row in csv_reader:
            rows.append(row)

    return rows
```

```python
rows = load_dataset('customer_support_tickets.csv')
print(rows[0])
```

```python
def get_unique_ticket_types(rows):
    if rows is None or len(rows) == 0:
        return set()
    unique_ticket_types = set()
    for row in rows:
        unique_ticket_types.add(row['Ticket Type'])

    return unique_ticket_types
```

```python
def get_unique_statuses(rows):
    if rows is None or len(rows) == 0:
        return set()
    unique_statuses = set()
    for row in rows:
        unique_statuses.add(row['Ticket Status'])
    return unique_statuses
```

```python
def get_unique_priorities(rows):
    if rows is None or len(rows) == 0:
        return set()
    unique_priorities = set()
    for row in rows:
        unique_priorities.add(row['Ticket Priority'])

    return unique_priorities
```

```python
def get_unique_products(rows):
    if rows is None or len(rows) == 0:
        return set()
    unique_products = set()
    for row in rows:
        unique_products.add(row['Product Purchased'])

    return unique_products
```

```python
def get_total_tickets(rows):
    return len(rows)
```

```python
print(get_total_tickets(rows))
print(get_unique_ticket_types(rows))
print(get_unique_statuses(rows))
print(get_unique_priorities(rows))
print(get_unique_products(rows))
```

```python
# def find_tickets_by_product(rows, product):
#     matched_products = []
#     for row in rows:
#         if row['Product Purchased'] == product:
#             matched_products.append(row)
#
#     return matched_products

def find_tickets_by_product(rows, product):
    if rows is None or len(rows) == 0:
        return []

    if product is None or product == "":
        return []

    matched_products = []

    for row in rows:
        if row.get("Product Purchased") == product:
            matched_products.append(row)

    return matched_products
```

```python
def count_by_ticket_type(rows):
    frequency_ticket_type_dict = {}
    for row in rows:
        ticket_type = row['Ticket Type']
        if ticket_type not in frequency_ticket_type_dict:
            frequency_ticket_type_dict[ticket_type] = 1
        else:
            frequency_ticket_type_dict[ticket_type] += 1

    return frequency_ticket_type_dict

```

```python
def count_by_status(rows):
    frequency_status_dict = {}

    for row in rows:
        if row['Ticket Status'] not in frequency_status_dict:
            frequency_status_dict[row['Ticket Status']] = 1
        else :
            frequency_status_dict[row['Ticket Status']] += 1

    return frequency_status_dict
```

```python
def count_by_priority(rows):
    frequency_priority_dict = {}

    for row in rows:
        if row['Ticket Priority'] not in frequency_priority_dict:
            frequency_priority_dict[row['Ticket Priority']] = 1
        else :
            frequency_priority_dict[row['Ticket Priority']] += 1

    return frequency_priority_dict
```

```python
def count_by_product(rows):
    frequency_product_dict = {}

    for row in rows:
        if row.get('Product Purchased') not in frequency_product_dict:
            frequency_product_dict[row.get('Product Purchased')] = 1
        else :
            frequency_product_dict[row.get('Product Purchased')] += 1

    return frequency_product_dict
```

```python
print(count_by_ticket_type(rows))
print(count_by_status(rows))
print(count_by_priority(rows))
print(count_by_product(rows))
```

```python
def find_duplicate_ticket_ids(rows):
    unique_ticket_ids = set()
    duplicate_ticket_ids = set()

    for row in rows:
        ticket_id = row.get("Ticket ID")

        if ticket_id in unique_ticket_ids:
            duplicate_ticket_ids.add(ticket_id)
        else:
            unique_ticket_ids.add(ticket_id)

    return duplicate_ticket_ids
```

```python
def group_tickets_by_customer(rows):
    customer_tickets = {}

    for row in rows:
        customer = row.get("Customer Name")

        if customer not in customer_tickets:
            customer_tickets[customer] = [row]
        else:
            customer_tickets[customer].append(row)

    return customer_tickets
```

```python
def find_customers_with_multiple_tickets(customer_tickets):
    multiple_ticket_customers = []

    for customer, tickets in customer_tickets.items():
        if len(tickets) > 1:
            multiple_ticket_customers.append(customer)

    return multiple_ticket_customers
```

```python
def build_ticket_lookup(rows):
    ticket_lookup = {}
    for row in rows:
        ticket_lookup[row['Ticket ID']] = row
    return ticket_lookup
```

```python
def get_ticket_by_id(ticket_lookup, ticket_id):
    if ticket_id in ticket_lookup:
        return ticket_lookup[ticket_id]
    return None
```

```python
def top_k_ticket_types(rows, k):
    ticket_types_freq = {}
    for row in rows:
        if row['Ticket Type'] not in ticket_types_freq:
            ticket_types_freq[row['Ticket Type']] = 1
        else:
            ticket_types_freq[row['Ticket Type']] += 1

    sorted_ticket_type_freq = sorted(ticket_types_freq.items(), key=lambda item: item[1], reverse=True)

    return dict(sorted_ticket_type_freq[:k])
```

```python
# =========================
# Assignment 4 - Results
# =========================

print("=" * 60)
print("ASSIGNMENT 4: CUSTOMER SUPPORT TICKET ANALYZER")
print("=" * 60)

# Part 1: Basic Exploration
print("\n[Part 1] BASIC EXPLORATION")
print("-" * 40)

print("Total tickets:", get_total_tickets(rows))
print("Unique ticket types:", get_unique_ticket_types(rows))
print("Unique statuses:", get_unique_statuses(rows))
print("Unique priorities:", get_unique_priorities(rows))

unique_products = get_unique_products(rows)
print("Number of unique products:", len(unique_products))
print("Unique products:", unique_products)


# Part 2: Searching
print("\n[Part 2] SEARCHING")
print("-" * 40)

customer_name = "Marisa Obrien"
customer_results = find_tickets_by_customer(rows, customer_name)

print(f"Tickets for '{customer_name}':", len(customer_results))

if customer_results:
    print("First matching ticket:")
    print(customer_results[0])


product_name = "GoPro Hero"
product_results = find_tickets_by_product(rows, product_name)

print(f"\nTickets for product '{product_name}':", len(product_results))


# Part 3: Frequency Counting
print("\n[Part 3] FREQUENCY COUNTING")
print("-" * 40)

ticket_type_counts = count_by_ticket_type(rows)
status_counts = count_by_status(rows)
priority_counts = count_by_priority(rows)
product_counts = count_by_product(rows)

print("Ticket type frequencies:")
print(ticket_type_counts)

print("\nStatus frequencies:")
print(status_counts)

print("\nPriority frequencies:")
print(priority_counts)

print("\nTop 10 products by ticket count:")
for product, count in sorted(
    product_counts.items(),
    key=lambda item: item[1],
    reverse=True
)[:10]:
    print(f"{product}: {count}")


# Part 4: Duplicate Detection
print("\n[Part 4] DUPLICATE TICKET IDS")
print("-" * 40)

duplicate_ticket_ids = find_duplicate_ticket_ids(rows)

print("Number of duplicate ticket IDs:", len(duplicate_ticket_ids))
print("Duplicate ticket IDs:", duplicate_ticket_ids)


# Part 5: Customer Grouping
print("\n[Part 5] CUSTOMER GROUPING")
print("-" * 40)

customer_tickets = group_tickets_by_customer(rows)

print("Number of unique customers:", len(customer_tickets))

multiple_ticket_customers = find_customers_with_multiple_tickets(
    customer_tickets
)

print(
    "Customers with multiple tickets:",
    len(multiple_ticket_customers)
)

print("\nSample customers with multiple tickets:")

for customer in multiple_ticket_customers[:10]:
    print(
        f"{customer}: "
        f"{len(customer_tickets[customer])} tickets"
    )


# Part 6: Fast Ticket Lookup
print("\n[Part 6] FAST TICKET LOOKUP")
print("-" * 40)

ticket_lookup = build_ticket_lookup(rows)

ticket_id = "1"
ticket = get_ticket_by_id(ticket_lookup, ticket_id)

print(f"Lookup result for Ticket ID {ticket_id}:")

if ticket:
    print(ticket)
else:
    print("Ticket not found")


# Part 7: Top-K
print("\n[Part 7] TOP-K TICKET TYPES")
print("-" * 40)

k = 3
top_ticket_types = top_k_ticket_types(rows, k)

print(f"Top {k} ticket types:")

for ticket_type, count in top_ticket_types.items():
    print(f"{ticket_type}: {count}")


# Part 8: Complexity Summary
print("\n[Part 8] COMPLEXITY SUMMARY")
print("-" * 40)

print("Searching original list: O(n)")
print("Frequency counting: O(n)")
print("Finding unique values: O(n) expected")
print("Duplicate detection: O(n) average")
print("Customer grouping: O(n) average")
print("Building ticket lookup: O(n)")
print("Ticket lookup using dictionary: O(1) average")
print("Top-K sorting: O(n + k_unique log k_unique)")

print("\n" + "=" * 60)
print("ASSIGNMENT 4 COMPLETE")
print("=" * 60)
```
