```python
class BankAccount():
  account_number = 1001
  def __init__(self , name , amount):

    self.name = name
    self.acc_no = BankAccount.account_number
    BankAccount.account_number += 1
    self.__amount = amount
    print(f"{self.name} 's New account opened")


  def __repr__(self):
    return f"{self.name} with account no {self.acc_no} has INR {self.__amount}"

  def get_balance(self):
    return f"{self.name} has INR {self.__amount}"

  def withdrawl(self , amt):
    if amt > self.__amount:
      print("Insufficient Funds")
      return
    if amt == 0 or amt < 0:
      print("Invalid Amount")
      return
    self.__amount -= amt
    print(f"{self.name} has withdrawn INR {amt}")

  def deposit(self , amt):
    if amt <= 0:
      print("Invalid Amount")
      return
    self.__amount += amt
    print(f"{self.name} has deposited INR {amt} \n the new balance is INR {self.__amount}")

  def __add__(self , other):
    new_name = f"{self.name} + {other.name}"
    new_amount = self.__amount + other.__amount
    return BankAccount(new_name , new_amount)

```

```python
aay_Acc = BankAccount("Aayush " ,12000)
moh_Acc = BankAccount("Mohit " ,1200)
hag_Acc = BankAccount("Hagmor" ,12033)
```

```python
joint_Acc = aay_Acc + moh_Acc
print(joint_Acc)
```

```python
class saving_account(BankAccount):
  def __init__(self, name, amount):
    super().__init__(name, amount)
    self.interest_rate = 0.035
    self.__amount = amount

  def calculate_interest(self):
    interest = self.__amount * self.interest_rate
    self.__amount += interest
    print(f"{self.name} has earned INR {interest :.2f} and now the balance is INR {self.__amount: .2f}")
```

```python
aay_Acc = saving_account("Aayush " ,12000)
moh_Acc = saving_account("Mohit " ,1200)
hag_Acc = saving_account("Hagmor" ,12033)
```

```python
aay_Acc.calculate_interest()
moh_Acc.calculate_interest()
hag_Acc.calculate_interest()
```
